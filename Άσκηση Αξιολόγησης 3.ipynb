```python
import sympy as sp
from IPython.display import display

x, y = sp.symbols('x y')        # these are the only two variables used

class WaveFunctions:                    # the first class contains only the functions
    psi = sp.sin(2 * x) * sp.cos(5 * x)     # the functions take this form in code
    phi = sp.exp(-2 * (x**2 + y**2))        # they have been checked by hand if they are eigenfunctions
    chi = sp.exp(-sp.I * (x + y))           # and indeed, one of them is

class Operators:                                            # the second class contains the operators
    def A(f): return sp.diff(f, x) + sp.diff(f, y)              # this takes a function f, and differentiates it according to A
    def B(f): return sp.diff(f, x, 2) + sp.diff(f, y, 2) + 1    # derivatives of the second order are implemented with a 2 in diff

class EigenfunctionChecker:                         # the third class does everything
    def __init__(self, operators, wave_functions):      # the init function takes the previous two classes as "input"
        self.operators = operators
        self.wave_functions = wave_functions
    
    def eigencheck(self, op, f):
        ratio = sp.simplify(op(f) / f)                                                  # if f is an eigenfunction, ratio should not contain x or y
        is_constant = (x not in ratio.free_symbols) and (y not in ratio.free_symbols)   # ratio.free_symbols checks for that
        return ratio if is_constant else None                                           # is_constant is a boolean
    
    def commutation(self, f):
        AB = Operators.A(Operators.B(f))    # first apply B, then A
        BA = Operators.B(Operators.A(f))
        return AB, BA, AB - BA

    def run_all(self):
        functions = {"ψ": self.wave_functions.psi, "φ": self.wave_functions.phi, "χ": self.wave_functions.chi}  # this line names the functions
        operators = {"A": self.operators.A, "B": self.operators.B}                                      # this line gives names to the operators
        for fname, f in functions.items():                                              # these loops run for every function, and every operator
            print(f"{fname}(x, y) = {f}")
            for oname, op in operators.items():         # the fname and oname are needed only for printing the results
                eigenvalue = self.eigencheck(op, f)         # this is either the eigenvalue or nothing at all
                if eigenvalue: print(f"  {oname}({fname}): {'eigenfunction with λ = ' + str(eigenvalue)}")  # str(eigenvalue) works because of sympy
                else: print(f"  {oname}({fname}): {'not a eigenfunction'}")

            com_ab, com_ba, com_result = self.commutation(f)    # for every function, calculate the commutator
            print(f"    AB({fname}) = {sp.simplify(com_ab)}")
            print(f"    BA({fname}) = {sp.simplify(com_ba)}")
            print(f"      [A, B]({fname}) = {sp.simplify(com_result)}\n")     # simplify is used to make the expressions smaller

if __name__ == "__main__": EigenfunctionChecker(Operators, WaveFunctions).run_all()     # only this line is required to run the entire code
```
